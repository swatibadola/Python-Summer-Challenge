## Question 1

What is the average click-through rate (CTR) for sponsored product ads for each product category that contains the substring 'Electronics' in its name during October 2024? This analysis will help determine which electronics-related categories are performing optimally.

```python
# Note: pandas and numpy are already imported as pd and np
# The following tables are loaded as pandas DataFrames with the same names: fct_ad_performance, dim_product
# Please print your final result or dataframe

def get_avg_ctr(fct_ad_performance, dim_product):
  # merging tables on product_id
  merged = pd.merge(fct_ad_performance, dim_product, on='product_id', how='inner')

  #filter for october 2024
  merged['recorded_date'] = pd.to_datetime(merged['recorded_date'])
  october_2024 = merged[
    (merged['recorded_date'].dt.year == 2024) &
    (merged['recorded_date'].dt.month == 10)
  ]

  # filter for product_category containing 'electronics'
  electronics_data = october_2024[
    october_2024['product_category'].str.contains('Electronics', case=False, na=False)
  ]

  # calculating ctr
  electronics_data = electronics_data[electronics_data['impressions'] > 0]
  electronics_data['ctr'] = electronics_data['clicks'] / electronics_data['impressions']

  # grouping by category and calc average ctr
  result = electronics_data.groupby('product_category')['ctr'].mean().reset_index()
  result.rename(columns={'ctr': 'average_ctr'}, inplace=True)

  return result

print(get_avg_ctr(fct_ad_performance, dim_product))
```

## Question 2

Which product categories have a CTR greater than the aggregated overall average CTR for sponsored product ads during October 2024? This analysis will identify high-performing categories for further optimization. For this question, we want to calculate CTR for each ad, then get the average across ads by product category & overall.

```python
def ctr_for_each_ad(fct_ad_performance, dim_product):
  df = pd.merge(fct_ad_performance, dim_product, on='product_id', how='inner')

  #filter for october 2024
  df['recorded_date'] = pd.to_datetime(df['recorded_date'])
  df_october_2024 = df[
    (df['recorded_date'].dt.year == 2024) &
    (df['recorded_date'].dt.month == 10)
  ]

  # remove rows with 0 impressions (to avoid division by 0)
  df_october_2024 = df_october_2024[df_october_2024['impressions'] > 0]

  # calculating ctr per ad
  df_october_2024['ctr'] = df_october_2024['clicks'] / df_october_2024['impressions']

  # avg ctr per product_category
  category_ctr = df_october_2024.groupby('product_category')['ctr'].mean().reset_index()
  category_ctr.rename(columns={'ctr' : 'average_ctr'}, inplace=True)

  # calc overall avg ctr across all ads
  overall_avg_ctr = df_october_2024['ctr'].mean()

  # filtering categories
  high_performing = category_ctr[category_ctr['average_ctr'] > overall_avg_ctr]

  return high_performing

print(ctr_for_each_ad(fct_ad_performance, dim_product))
```

## Question 3

For the product categories identified in the previous question, what is the percentage difference between their CTR and the overall average CTR for October 2024? This analysis will quantify the performance gap to recommend specific categories for targeted advertising optimization.

```python
def performance_gap(fct_ad_performance, dim_product):
  df = pd.merge(fct_ad_performance, dim_product, on='product_id', how='inner')

  # filter for oct 2024
  df['recorded_date'] = pd.to_datetime(df['recorded_date'])
  df_oct_2024 = df[
    (df['recorded_date'].dt.year == 2024) &
    (df['recorded_date'].dt.month == 10)
  ]

  # removing rows with 0 impressions
  df_oct_2024 = df_oct_2024[df_oct_2024['impressions'] > 0]

  # calc ctr per ad 
  df_oct_2024['ctr'] = df_oct_2024['clicks'] / df_oct_2024['impressions']

  # avg ctr per product_category
  category_ctr = df_oct_2024.groupby('product_category')['ctr'].mean().reset_index()
  category_ctr.rename(columns = {'ctr' : 'average_ctr'}, inplace=True)


  # calc overall avg ctr accross all ads
  overall_avg_ctr = df_oct_2024['ctr'].mean()

  # filtering categories
  high_performing = category_ctr[category_ctr['average_ctr'] > overall_avg_ctr].copy()

  # calc % diff from overall ctr
  high_performing['percentage_difference'] = (
    (high_performing['average_ctr'] - overall_avg_ctr) / overall_avg_ctr) * 100

  return high_performing.sort_values(by='percentage_difference', ascending=False)

print(performance_gap(fct_ad_performance, dim_product))
```
